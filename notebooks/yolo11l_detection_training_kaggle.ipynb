```python
from ultralytics import YOLO
import yaml
from pathlib import Path

DATASET_ROOT = Path("/kaggle/input/datasets/manuelht311/roboflow-football-detection")
model = YOLO('yolo11l.pt')

results = model.train(
    data='/kaggle/input/datasets/manuelht311/roboflow-football-detection/data.yaml',
    epochs=50,
    patience=20,
    batch=6,
    imgsz=1280,
    plots=True,
    device=[0,1],
    project='soccernet_finetune',
    name='yolo11l_roboflow_1280'
)
```
